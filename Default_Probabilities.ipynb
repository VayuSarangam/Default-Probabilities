```python
import numpy as np
import pandas as pd

n_loans = 1000
rng = np.random.default_rng(1000)

property_types = np.array(["Industrial", "Multifamily", "Retail", "Office", "Hotel"])


property_risk_map = {"Industrial": 0.00, "Multifamily": 0.20, "Retail": 0.35, "Office": 0.85, "Hotel": 1.00}

loan_tape = pd.DataFrame({ "Loan id": [f"LN-{i:04d}" for i in range(1, n_loans + 1)], "Property type": rng.choice(property_types, size=n_loans), "LTV": rng.uniform(0.40, 0.95, size=n_loans),
    "DSCR": rng.uniform(0.75, 2.50, size=n_loans),
    "Occupancy": rng.uniform(0.50, 1.00, size=n_loans),
    "Sponsor risk score": rng.integers(1, 6, size=n_loans),
    "Market stress index": rng.uniform(0.00, 1.00, size=n_loans),
    "Loan age years": rng.uniform(0.00, 10.00, size=n_loans),
})

loan_tape["Property Risk Flag"] = loan_tape["Property type"].map(property_risk_map)

loan_tape.head(200).style.format({
    "LTV": "{:.2%}",
    "DSCR": "{:.2f}",
    "Occupancy": "{:.2%}",
    "Market stress index": "{:.2f}",
    "Property Risk Flag": "{:.2f}"
})


print (loan_tape.shape)
print (loan_tape.describe())
print (loan_tape["Property type"].value_counts())

noise = rng.normal(loc=0.0, scale=0.40, size=n_loans)

risk_score = (
    -2.50
    + 3.00 * (loan_tape["LTV"] - 0.65)
    - 1.10 * (loan_tape["DSCR"] - 1.25)
    - 2.20 * (loan_tape["Occupancy"] - 0.85)
    + 0.30 * (loan_tape["Sponsor risk score"] - 3)
    + 0.75 * loan_tape["Property Risk Flag"]
    + 1.10 * (loan_tape["Market stress index"] - 0.50)
    + 0.03 * (loan_tape["Loan age years"] - 5)
    + noise
)

Def_prob = 1/(1+np.exp(-risk_score))

print()
print("Default Probability:")
print(Def_prob.head(10).round(2)*100, "%")
```
